```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.datasets import face

import fsc
```

```python
image = face(gray=True).astype(float)
plt.imshow(image, cmap=plt.cm.gray)
```

```python
plt.ylim(-1.1, 1.1)
plt.plot(*fsc.compute_fsc(image,vox_size=(1.,1.), bins=100, full=True))
```

```python
noiseimage=image+np.random.randn(*image.shape)*20
plt.imshow(noiseimage, cmap=plt.cm.gray)
```

```python
plt.ylim(-1.1, 1.1)
plt.plot(*fsc.compute_fsc(noiseimage,vox_size=(1.,1.), bins=500, full=True))
```

```python
plt.ylim(-1.1, 1.1)
plt.plot(*fsc.fsc2(noiseimage,image,vox_size=(1.,1.), bins=500, full=True))
```
